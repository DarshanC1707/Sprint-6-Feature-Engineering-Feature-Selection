# Sprint 6 – Notebook 9 – Feature Selection



```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
rng = np.random.default_rng(42)
n = 120
customers = pd.DataFrame({
    "Customer_ID": np.arange(1001, 1001+n),
    "Age": rng.integers(19, 65, n),
    "Annual_Income": rng.integers(250000, 1400000, n),
    "Membership_Type": rng.choice(["Basic", "Silver", "Gold"], n),
    "City": rng.choice(["Hassan", "Bengaluru", "Mysuru"], n),
    "Product_Category": rng.choice(["Electronics", "Clothing", "Grocery"], n),
    "Quantity": rng.integers(1, 7, n),
    "Product_Price": rng.integers(300, 9000, n),
    "Discount_Percent": rng.choice([0, 5, 10, 15, 20], n),
    "Purchase_Date": pd.date_range("2026-01-01", periods=n, freq="D"),
})
customers["Total_Spending"] = (customers.Quantity * customers.Product_Price * (1-customers.Discount_Percent/100)).round(2)
score = (customers["Membership_Type"].eq("Gold").astype(int)*1.4 + customers["Total_Spending"]/20000 + customers["Age"]/90 + rng.normal(0,0.7,n))
customers["Repeat_Purchase"] = (score > np.median(score)).astype(int)
display(customers.head())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectKBest, f_classif, VarianceThreshold, mutual_info_classif, chi2, RFE, SelectFromModel
X = pd.get_dummies(customers[["Age","Annual_Income","Membership_Type","City","Product_Category","Quantity","Product_Price","Discount_Percent"]], dtype=int)
y = customers["Repeat_Purchase"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=.25, random_state=42, stratify=y)
print(X_train.shape, X_test.shape)
```
