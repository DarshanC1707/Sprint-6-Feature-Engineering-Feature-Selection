# Sprint 6 – Notebook 1: Introduction to Feature Engineering


```python
import pandas as pd
import numpy as np

customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Age": [22, 35, 28, 45, 31, 52, 26, 39],
    "Annual_Income": [300000, 650000, 450000, 900000, 550000, 1100000, 380000, 720000],
    "City": ["Hassan", "Bengaluru", "Mysuru", "Bengaluru", "Hassan", "Mysuru", "Hassan", "Bengaluru"],
    "Membership_Type": ["Basic", "Gold", "Basic", "Gold", "Silver", "Gold", "Silver", "Basic"],
    "Product_Category": ["Electronics", "Clothing", "Grocery", "Electronics", "Clothing", "Electronics", "Grocery", "Clothing"],
    "Product_Price": [15000, 2000, 800, 25000, 3000, 40000, 1200, 3500],
    "Quantity": [1, 3, 5, 2, 2, 1, 4, 3],
    "Discount_Percent": [5, 10, 0, 15, 5, 20, 0, 10],
    "Purchase_Date": pd.to_datetime(["2026-01-12", "2026-02-03", "2026-02-14", "2026-03-05", "2026-03-18", "2026-04-02", "2026-04-10", "2026-04-22"]),
    "Repeat_Purchase": [0, 1, 0, 1, 1, 1, 0, 1]
})
customers.head()
```

## 1. What is a Feature?

### Definition
A feature is an input value used by a machine learning model to make a prediction. Features describe measurable properties of an observation. They may be numerical, categorical, or derived from dates. Each row supplies feature values for one example.

### Business Example
An online store records each shopper’s age, city, and purchase price. These columns describe a customer or transaction. The business uses them to understand shopping behavior. Each column can be considered as a possible feature.

### AI/ML Use
A model uses feature values to learn relationships with an outcome. For repeat-purchase prediction, age and spending may be useful inputs. Features must be available when the prediction is made. Better input information can help a model learn useful patterns.

### Python Example

```python
print(customers[["Age", "City", "Product_Price"]].head())
```

### Explanation
The code selects three input columns. Each row contains one customer’s feature values. The target column is intentionally excluded. These columns could later be passed to a model.

## 2. Feature vs Variable

### Definition
A variable is a general name for a value that can change. A feature is specifically an input used for analysis or prediction. A dataset variable is not automatically a useful model feature. Identifiers and target columns may be variables but should not be treated as ordinary predictive inputs.

### Business Example
A retailer stores Customer_ID, Age, and Repeat_Purchase. All three are variables in the dataset. Age may be used as a predictive feature. Customer_ID is usually an identifier, while Repeat_Purchase is the outcome.

### AI/ML Use
Distinguishing variables from features prevents accidental inclusion of irrelevant fields. Unique identifiers often provide little generalizable signal. A target must be separated from inputs. The model should use meaningful information rather than memorize records.

### Python Example

```python
print("All variables:", customers.columns.tolist())
print("Example feature:", customers["Age"].name)
```

### Explanation
The first line lists every variable in the dataset. The second line names one potential feature. Not every listed variable should be given to a model. The modeling goal determines which variables become features.

## 3. Feature vs Target

### Definition
Features are the information provided to a model. The target is the value the model is trained to predict. In supervised learning, the target is separated from the input features. The target must not be included in the input matrix.

### Business Example
An e-commerce team wants to predict repeat purchases. Customer details and transaction values are possible inputs. Repeat_Purchase is the desired outcome. Keeping these roles separate makes the prediction task clear.

### AI/ML Use
Machine learning libraries often call inputs X and targets y. The model learns a relationship from X to y during training. Putting y into X creates direct leakage. A valid evaluation requires the target to remain separate.

### Python Example

```python
X = customers.drop(columns=["Repeat_Purchase", "Customer_ID"])
y = customers["Repeat_Purchase"]
print("Features:", X.shape[1], "Target:", y.name)
```

### Explanation
The code removes the target and identifier from X. It stores the outcome in y. X contains candidate inputs for prediction. This is a basic supervised-learning setup.

## 4. What is Feature Engineering?

### Definition
Feature engineering is the process of creating, modifying, or organizing input features. It turns raw data into information that a model can use more effectively. The process can include creating ratios, grouping categories, and extracting dates. It should be guided by the prediction goal and available data.

### Business Example
A shop stores product price, quantity, and discount separately. The team can calculate the final order amount. That new value is easier to interpret as transaction spending. It may be more informative than price alone.

### AI/ML Use
Feature engineering helps models detect patterns that are difficult to learn from raw columns. It can improve signal quality and interpretability. It does not guarantee better accuracy in every case. Engineered features must be evaluated on held-out data.

### Python Example

```python
orders = customers.copy()
orders["Order_Value"] = orders["Product_Price"] * orders["Quantity"] * (1 - orders["Discount_Percent"] / 100)
print(orders[["Product_Price", "Quantity", "Order_Value"]].head())
```

### Explanation
The code combines price, quantity, and discount. Order_Value estimates the amount paid for an order. It adds a meaningful numerical feature. The original columns remain available for comparison.

## 5. Why Feature Engineering?

### Definition
Raw data may not express useful patterns directly. Feature engineering can reveal relationships through combinations or transformations. It can also make data more suitable for particular algorithms. The goal is to improve usefulness, not simply increase the number of columns.

### Business Example
A store knows purchase dates but wants to understand weekend demand. A weekend indicator directly captures this behavior. Staff can compare weekend and weekday purchases. This is often easier than interpreting raw dates.

### AI/ML Use
Engineered features may help a model identify seasonality, customer segments, or spending behavior. They can reduce the burden on simple algorithms. However, unnecessary features may introduce noise. Every new feature should have a clear reason for inclusion.

### Python Example

```python
orders["Is_Weekend"] = (orders["Purchase_Date"].dt.dayofweek >= 5).astype(int)
print(orders[["Purchase_Date", "Is_Weekend"]].head())
```

### Explanation
The code converts a date into a weekend flag. One means Saturday or Sunday, and zero means a weekday. This is easier for many models to interpret than a raw timestamp. It demonstrates why feature engineering can be useful.

## 6. Importance of Domain Knowledge

### Definition
Domain knowledge means understanding how a particular business or process works. It helps choose features that reflect meaningful real-world relationships. It also helps identify unrealistic values and leakage risks. Domain knowledge should guide feature ideas rather than replace testing.

### Business Example
In retail, order value depends on price, quantity, and discounts. A business analyst knows that high prices do not always mean high order totals. This insight suggests calculating a net order value. The new feature reflects how purchases actually work.

### AI/ML Use
Domain knowledge can suggest features before any model is trained. It helps avoid nonsensical ratios or misleading categories. It is especially important when data represents financial or operational processes. The model still needs validation to confirm usefulness.

### Python Example

```python
orders["Gross_Value"] = orders["Product_Price"] * orders["Quantity"]
print(orders[["Gross_Value", "Order_Value"]].head())
```

### Explanation
Gross_Value represents spending before discounts. Order_Value represents spending after discounts. The comparison follows a real retail business rule. This is an example of domain knowledge guiding feature creation.

## 7. Good Features vs Bad Features

### Definition
A good feature is relevant, reliable, and available at prediction time. It should capture information that may generalize to new examples. A bad feature may be noisy, redundant, misleading, or leaked from the future. Feature quality matters more than feature count.

### Business Example
Age and purchase amount may be available when a customer checks out. A customer ID is mainly used to identify a record. A future purchase confirmation is not available at checkout. That future value would be a bad feature for predicting repeat purchase.

### AI/ML Use
Useful features can improve model learning and interpretation. Irrelevant features may encourage overfitting. Leakage features can produce unrealistically strong evaluation scores. Feature choices should be reviewed against the intended prediction moment.

### Python Example

```python
print("Useful candidate:", orders["Order_Value"].name)
print("Identifier (usually exclude):", orders["Customer_ID"].name)
```

### Explanation
The code contrasts a meaningful transaction value with an identifier. Order_Value captures spending behavior. Customer_ID mainly labels records. Whether a feature is good depends on the task and when it becomes available.

## 8. Feature Engineering Workflow

### Definition
A feature engineering workflow begins by defining the prediction problem. Next, inspect raw columns and identify possible improvements. Create or transform features using training data where fitting is required. Finally, evaluate the resulting features on separate validation data.

### Business Example
A retail team starts with a repeat-purchase objective. It checks customer, product, and date fields. It creates order value and weekend indicators. The team then checks whether those features improve predictions.

### AI/ML Use
A structured workflow reduces accidental leakage and inconsistent transformations. Train/test splitting should occur before fitting learned transformations. Reusable pipelines help apply the same logic to new records. Feature evaluation should compare against a baseline.

### Python Example

```python
print("1. Define target: Repeat_Purchase")
print("2. Inspect columns:", len(customers.columns))
print("3. Create features: Order_Value, Is_Weekend")
print("4. Validate on held-out data")
```

### Explanation
The code displays the major workflow stages. The target guides feature decisions. Raw data is inspected before transformations. Final features should be checked on data not used to learn preprocessing parameters.

## 9. Feature Creation

### Definition
Feature creation combines existing information into new columns. Common examples include totals, ratios, counts, and flags. New features should have a clear interpretation. Avoid creating values that depend on information unavailable at prediction time.

### Business Example
A shop can calculate the discount amount for each purchase. This amount is not directly stored in the raw dataset. It is derived from price, quantity, and discount percentage. The business can use it to study promotional spending.

### AI/ML Use
New features can expose meaningful relationships between raw variables. A model may find total discount more informative than discount percentage alone. Creating too many combinations can add noise. The value of each feature must be tested.

### Python Example

```python
orders["Discount_Amount"] = orders["Gross_Value"] - orders["Order_Value"]
print(orders[["Gross_Value", "Discount_Amount"]].head())
```

### Explanation
The code subtracts the net order value from the gross value. The result is the discount amount in currency units. This new feature describes the transaction directly. It is created without learning parameters from the target.

## 10. Feature Transformation

### Definition
Feature transformation changes the representation or scale of an existing column. Examples include logarithms, standardization, and category encoding. The underlying information is reorganized rather than replaced by an unrelated measure. Some transformations must be fitted on training data only.

### Business Example
Retail spending amounts can vary widely across customers. A logarithmic transformation can reduce the influence of very large values. This may make spending patterns easier to model. It does not mean the original purchases were wrong.

### AI/ML Use
Transformations can help scale-sensitive algorithms and skewed numerical distributions. Log transforms require careful handling of zeros and negatives. Scalers should learn their parameters from training data. Test data must use those same learned parameters.

### Python Example

```python
orders["Log_Order_Value"] = np.log1p(orders["Order_Value"])
print(orders[["Order_Value", "Log_Order_Value"]].head())
```

### Explanation
The code uses log1p, which calculates log(1 + value). It compresses large positive values while preserving ordering. It works for zero and positive order amounts. The transformed column can be compared with the original.

## 11. Feature Selection

### Definition
Feature selection chooses a subset of available features. It removes inputs that are irrelevant, redundant, or unsuitable. Selection can use domain knowledge, statistical methods, or model-based methods. It should be evaluated without using test-set information.

### Business Example
A retailer may store Customer_ID, Age, City, and Order_Value. Customer_ID usually should not be a model input. Age, City, and Order_Value may carry useful information. The team keeps candidates relevant to its prediction goal.

### AI/ML Use
Selecting useful features can reduce noise and model complexity. It may improve interpretability and training speed. A feature that looks weak alone can still help in combination with others. Automated selection should be fitted on training data only.

### Python Example

```python
selected = orders[["Age", "City", "Order_Value"]]
print(selected.head())
```

### Explanation
The code creates a smaller set of candidate features. It excludes the identifier and target. The selection is illustrative rather than a proven optimal subset. Model validation is needed to confirm which features help.

## 12. Feature Extraction

### Definition
Feature extraction converts raw information into a different set of informative attributes. It is often used for text, images, dates, and high-dimensional data. Extraction can produce several new values from one source. Unlike simple selection, it creates a new representation.

### Business Example
An online shop stores the date of each purchase. The business extracts month and day of the week. These values make seasonal and weekly patterns easier to examine. The raw date remains available for auditing.

### AI/ML Use
Extracted date components can help a model learn periodic behavior. More advanced extraction includes text vectorization and principal component analysis. Some extraction methods learn parameters and require training-only fitting. Extracted features must be available at prediction time.

### Python Example

```python
orders["Purchase_Month"] = orders["Purchase_Date"].dt.month
orders["Purchase_Weekday"] = orders["Purchase_Date"].dt.dayofweek
print(orders[["Purchase_Month", "Purchase_Weekday"]].head())
```

### Explanation
The code extracts two numerical values from each date. Month ranges from 1 to 12. Weekday ranges from 0 for Monday to 6 for Sunday. These values can reveal shopping patterns across time.

## 13. Feature Leakage

### Definition
Feature leakage happens when an input contains information unavailable at the time of prediction. It can also happen when preprocessing learns from validation or test data. Leakage often makes model performance look better than it really is. Avoiding leakage is essential for reliable evaluation.

### Business Example
Suppose a retailer predicts whether a customer will purchase again. A field created after the next purchase would reveal the answer. Using that field during training would be unfair. The model could not access it for a new customer at prediction time.

### AI/ML Use
Leaked features can cause misleadingly high validation scores. A safe workflow separates training and test data before fitting learned preprocessing. Feature availability must be checked against the prediction moment. Target-derived features must not enter the input matrix.

### Python Example

```python
safe_features = orders.drop(columns=["Repeat_Purchase", "Customer_ID"])
print("Target excluded:", "Repeat_Purchase" not in safe_features.columns)
print("Identifier excluded:", "Customer_ID" not in safe_features.columns)
```

### Explanation
The code excludes the target and customer identifier from the inputs. This avoids direct target leakage through the label column. Other forms of leakage still require careful review. In particular, future information must never be used to predict the past.
