## Dataset Setup

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
 "Customer_ID":[101,101,102,102,103,103,104,104],
 "Product_ID":["P1","P2","P1","P3","P2","P3","P1","P2"],
 "Age":[22,22,35,35,47,47,29,29],
 "Annual_Income":[300000,300000,650000,650000,900000,900000,450000,450000],
 "Product_Price":[1000,2500,1000,4000,2500,4000,1000,2500],
 "Quantity":[2,1,3,1,2,2,1,3],
 "Product_Rating":[4.2,4.8,4.2,4.5,4.8,4.5,4.2,4.8],
 "Purchase_Date":pd.to_datetime(["2026-01-03","2026-02-10","2026-01-09","2026-03-15","2026-02-04","2026-04-12","2026-01-20","2026-05-10"]),
 "Membership_Type":["Basic","Basic","Silver","Silver","Gold","Gold","Basic","Basic"],
 "Repeat_Purchase":[1,1,1,1,1,1,0,0]
})
df["Order_Value"] = df["Product_Price"] * df["Quantity"]
print(df.head())
```

## 1. Count

### Definition
Counts records in a group, such as orders per customer. It is a frequency measure, not a monetary total. This feature-engineering concept should be evaluated against the prediction goal.

### Business Example
A retailer counts orders for each customer to understand activity. The example uses customer shopping transactions to make the idea concrete.

### AI/ML Use
Purchase counts help predict repeat purchases and customer churn. Always check whether the feature is available at prediction time.

### Python Example

```python
print(df.groupby("Customer_ID").size().rename("Purchase_Count"))
```

### Explanation
Each customer ID is grouped and the number of transaction rows is counted. This is a compact example using the same sample dataset.

### When to Use
Use for event frequency. Choose this technique when it matches the data and business goal.

### Limitations
Multiple line items in one order should not be mistaken for multiple orders. Check for leakage and test whether the engineered feature actually improves validation performance.

## 8. Group-Based Aggregation

### Definition
Summarizes records based on one or more shared attributes. Different aggregations can be calculated together. This feature-engineering concept should be evaluated against the prediction goal.

### Business Example
A business compares spending and orders by membership tier. The example uses customer shopping transactions to make the idea concrete.

### AI/ML Use
Group summaries can create useful customer-segment signals. Always check whether the feature is available at prediction time.

### Python Example

```python
print(df.groupby("Membership_Type").agg(Orders=("Order_Value","size"), Sales=("Order_Value","sum")))
```

### Explanation
Two summary columns are calculated for each membership type. This is a compact example using the same sample dataset.

### When to Use
Use to compare segments. Choose this technique when it matches the data and business goal.

### Limitations
Segment-level features may leak future data if computed after prediction time. Check for leakage and test whether the engineered feature actually improves validation performance.

## 9. Customer-Level Aggregation

### Definition
Combines transaction rows into one row per customer. It creates a customer profile from purchase history. This feature-engineering concept should be evaluated against the prediction goal.

### Business Example
A marketing team summarizes customer lifetime orders and spend. The example uses customer shopping transactions to make the idea concrete.

### AI/ML Use
Customer-level metrics help predict retention and spending. Always check whether the feature is available at prediction time.

### Python Example

```python
customer = df.groupby("Customer_ID").agg(Total_Spending=("Order_Value","sum"), Number_of_Purchases=("Order_Value","size"), Average_Order_Value=("Order_Value","mean"))
print(customer)
```

### Explanation
Each customer receives spending, frequency, and average order features. This is a compact example using the same sample dataset.

### When to Use
Use for customer-level prediction. Choose this technique when it matches the data and business goal.

### Limitations
Only aggregate transactions available before the prediction date. Check for leakage and test whether the engineered feature actually improves validation performance.

## Why Aggregation Features Are Useful
Aggregation turns many transaction rows into concise customer or product summaries. It captures frequency, typical spend, total value, and variability. **Important:** for real prediction tasks, compute features only from historical transactions available before the prediction date; avoid joining future purchases or outcomes into training features.
