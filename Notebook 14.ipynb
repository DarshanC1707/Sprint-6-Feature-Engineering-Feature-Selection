# Sprint 6 – Notebook 13 - Domain-Based Feature Engineering

```python

import pandas as pd
import numpy as np

np.random.seed(42)

n = 100

df = pd.DataFrame({
    "person_age": np.random.randint(21, 65, n),
    "person_income": np.random.randint(20000, 150000, n),
    "person_home_ownership": np.random.choice(
        ["RENT", "OWN", "MORTGAGE", "OTHER"], n
    ),
    "person_emp_length": np.random.randint(0, 25, n).astype(float),
    "loan_intent": np.random.choice(
        ["EDUCATION", "MEDICAL", "VENTURE", "PERSONAL",
         "DEBTCONSOLIDATION", "HOMEIMPROVEMENT"], n
    ),
    "loan_grade": np.random.choice(list("ABCDEFG"), n),
    "loan_amnt": np.random.randint(1000, 30000, n),
    "loan_int_rate": np.round(np.random.uniform(5, 20, n), 2),
    "loan_status": np.random.choice([0, 1], n, p=[0.78, 0.22]),
    "loan_percent_income": np.round(
        np.random.uniform(0.02, 0.50, n), 2
    ),
    "cb_person_default_on_file": np.random.choice(["Y", "N"], n),
    "cb_person_cred_hist_length": np.random.randint(2, 30, n)
})

print("Dataset Shape:", df.shape)
print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Information:")
df.info()
```

## Topic 1 – Domain-Based Feature Engineering

### 1. Definition

Domain-Based Feature Engineering creates meaningful new features from existing data using business knowledge.

### 2. Business Example
Banks study customer income, loan amounts, and credit history to understand financial conditions.

### 3. AI/ML Use
- Credit Risk Prediction
- Loan Approval Prediction
- Fraud Detection
- Customer Segmentation

### 4. Python Example

```python
df["loan_income_ratio"] = (
    df["loan_amnt"] / df["person_income"]
)

print(df["loan_income_ratio"].head())
```

### 5. Explanation
Calculates the loan amount relative to annual income. It helps identify customers with higher borrowing burdens.

## Topic 2 – Identifying Raw Features

### 1. Definition

Raw Features are the original columns available in a dataset before creating new features.

### 2. Business Example

A bank collects customer age, income, employment length, loan amount, and credit history.

### 3. AI/ML Use

- Understanding Customer Profiles
- Identifying Important Variables
- Feature Selection
- Credit Risk Analysis

### 4. Python Example

```python
raw_features = [
    "person_age",
    "person_income",
    "person_emp_length",
    "loan_amnt",
    "loan_int_rate",
    "cb_person_cred_hist_length"
]

print(df[raw_features].head())
```

### 5. Explanation

- The list contains six original numerical features.
- `df[raw_features]` selects those columns from the dataset.
- These columns can be used to create meaningful new features.

## Topic 3 – Age Group Feature

### 1. Definition

Age Group Feature Engineering converts numerical age values into meaningful categories.

### 2. Business Example

Banks may study applicants in different age groups to understand customer profiles.

### 3. AI/ML Use

- Customer Segmentation
- Risk Analysis
- Customer Behaviour Analysis
- Credit Scoring

### 4. Python Example

```python
df["age_group"] = pd.cut(
    df["person_age"],
    bins=[0, 25, 35, 50, 100],
    labels=["Young", "Adult", "Middle_Aged", "Senior"]
)

print(df[["person_age", "age_group"]].head())
```

### 5. Explanation

- `pd.cut()` divides numerical age values into categories.
- Applicants are grouped into four age categories.
- This makes age information easier to analyse.
- Age-based features may be restricted or require careful fairness and legal review in lending applications.

## Topic 4 – Income Band Feature

### 1. Definition

Income Band Feature Engineering converts numerical income values into different income categories.

### 2. Business Example

Banks classify customers into income levels to understand their financial profiles.

### 3. AI/ML Use

- Credit Risk Prediction
- Customer Segmentation
- Loan Eligibility Analysis
- Financial Behaviour Analysis

### 4. Python Example

```python
df["income_band"] = pd.cut(
    df["person_income"],
    bins=[0, 30000, 60000, 100000, np.inf],
    labels=["Low", "Medium", "High", "Very_High"]
)

print(df[["person_income", "income_band"]].head())
```

### 5. Explanation

- `pd.cut()` divides annual income into four categories.
- Each applicant receives an income category.
- This helps identify differences between income groups.

## Topic 5 – Employment Stability Feature

### 1. Definition

Employment Stability Feature Engineering groups employment length into categories based on years of experience.

### 2. Business Example

Banks examine how long an applicant has been employed to understand employment history.

### 3. AI/ML Use

- Credit Risk Prediction
- Customer Profile Analysis
- Loan Assessment
- Financial Stability Analysis

### 4. Python Example


```python
df["employment_stability"] = pd.cut(
    df["person_emp_length"],
    bins=[-1, 2, 5, 10, np.inf],
    labels=["Low", "Moderate", "Good", "High"]
)

print(df[["person_emp_length", "employment_stability"]].head())
```

### 5. Explanation

- `person_emp_length` represents employment duration in years.
- `pd.cut()` groups employment duration into four categories.
- The new feature makes employment history easier to analyse.
- Employment duration alone does not fully measure financial stability.

## Topic 6 – Employment-to-Age Ratio

### 1. Definition

Employment-to-Age Ratio measures the proportion of an applicant's age represented by their employment duration.

### 2. Business Example

A bank may compare an applicant's employment history with their age to understand their work experience.

### 3. AI/ML Use

- Credit Risk Prediction
- Employment Profile Analysis
- Customer Segmentation
- Financial Behaviour Analysis

### 4. Python Example

```python
df["employment_age_ratio"] = (
    df["person_emp_length"] / df["person_age"].replace(0, np.nan)
)

print(df["employment_age_ratio"].head())
```

### 5. Explanation

- `person_emp_length` represents employment duration.
- `person_age` represents the applicant's age.
- The division calculates employment duration relative to age.
- A higher value indicates a larger proportion of age spent in employment.

### 5. Explanation

## Topic 8 – Previous Default Flag

### 1. Definition

Previous Default Flag converts historical default information into numerical values.

### 2. Business Example

Banks examine whether an applicant has previously defaulted on a loan.

### 3. AI/ML Use

- Credit Risk Prediction
- Default Prediction
- Customer Risk Assessment
- Loan Evaluation

### 4. Python Example

```python
df["previous_default_flag"] = (
    df["cb_person_default_on_file"].map({"Y": 1, "N": 0})
)

print(
    df[["cb_person_default_on_file", "previous_default_flag"]].head()
)
```

### 5. Explanation

- `Y` represents a previous default.
- `N` represents no previous default.
- `.map()` converts Y into 1 and N into 0.
- Numerical values make the information easier for ML algorithms to process.

## Topic 9 – Loan Grade Score

### 1. Definition

Loan Grade Score converts categorical loan grades into numerical scores.

### 2. Business Example

Banks classify loans into grades such as A, B, C, D, E, F, and G to represent different risk levels.

### 3. AI/ML Use

- Credit Risk Prediction
- Loan Risk Analysis
- Credit Scoring
- Risk Segmentation

### 4. Python Example

```python
grade_mapping = {
    "A": 7,
    "B": 6,
    "C": 5,
    "D": 4,
    "E": 3,
    "F": 2,
    "G": 1
}

df["loan_grade_score"] = df["loan_grade"].map(grade_mapping)

print(df[["loan_grade", "loan_grade_score"]].head())
```

### 5. Explanation

- Each loan grade is assigned a numerical score.
- Grade A receives 7, while Grade G receives 1.
- `.map()` converts categorical grades into numerical values.
- The grade's meaning and availability at prediction time must be verified before using it in a model.

## Topic 13 – Estimated Annual Interest

### 1. Definition

Estimated Annual Interest calculates the estimated yearly interest amount using the loan amount and interest rate.

### 2. Business Example

Banks estimate the annual interest associated with a loan to understand its financial cost.

### 3. AI/ML Use

- Loan Cost Analysis
- Financial Risk Analysis
- Customer Segmentation
- Loan Assessment

### 4. Python Example

```python
df["estimated_annual_interest"] = (
    df["loan_amnt"] * df["loan_int_rate"] / 100
)

print(df["estimated_annual_interest"].head())
```

## Topic 16 – Leakage Risk Verification

### 1. Definition

Data Leakage occurs when information that would not be available at prediction time is used to train a machine learning model.

### 2. Business Example

A bank must not use the final loan default result to predict whether the same applicant will default.

### 3. AI/ML Use

- Model Validation
- Reliable Prediction
- Training Data Verification
- Preventing Overly Optimistic Results

### 4. Python Example

```python
target_column = "loan_status"

engineered_features = [
    "loan_income_ratio",
    "age_group",
    "income_band",
    "employment_stability",
    "employment_age_ratio",
    "credit_history_age_ratio",
    "previous_default_flag",
    "loan_grade_score",
    "home_ownership_indicator",
    "loan_purpose_group",
    "loan_size_band",
    "estimated_annual_interest",
    "interest_income_ratio"
]

print("Target Column:", target_column)
print("Target Included in Features:",
      target_column in engineered_features)
```

### 5. Explanation

- `loan_status` is the target variable.
- `engineered_features` contains the candidate feature names.
- The code checks whether the target column is directly included in the feature list.
- A False result confirms that the target is not explicitly included.
- This is only a basic check; every feature must also be reviewed for prediction-time availability and indirect leakage.

## Topic 17 – Feature Retain and Remove Decision

### 1. Definition

Feature Retain and Remove Decision identifies which engineered features should be considered for model development and which should be excluded or reviewed.

### 2. Business Example

Banks review customer information before deciding which variables are appropriate for credit risk prediction.

### 3. AI/ML Use

- Feature Selection
- Model Optimization
- Reducing Unnecessary Features
- Improving Model Reliability

### 4. Python Example

```python
retained_features = [
    "loan_income_ratio",
    "income_band",
    "employment_stability",
    "previous_default_flag",
    "loan_size_band"
]

review_features = [
    "age_group",
    "employment_age_ratio",
    "credit_history_age_ratio",
    "loan_grade_score",
    "home_ownership_indicator",
    "loan_purpose_group",
    "estimated_annual_interest",
    "interest_income_ratio"
]

print("Candidate Features:", retained_features)
print("\nFeatures Requiring Review:", review_features)
```

### 5. Explanation

- `review_features` contains features requiring additional checks.
- Features should not be permanently retained or removed based only on assumptions.
- Final decisions require fairness, leakage, business, and model validation checks.
