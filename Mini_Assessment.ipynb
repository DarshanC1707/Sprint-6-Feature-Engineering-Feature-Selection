## Topic 13 – What is feature importance?

### Python Code

## Topic 14 – Create five useful features for an e-commerce dataset.

### Question / Task
Create five new features from customer purchases.

### Python Code

```python
d = pd.DataFrame({"Price":[100,200],"Quantity":[2,3],"Discount":[10,20],"Age":[25,45]})
d["Gross"] = d.Price*d.Quantity
d["Discount_Amount"] = d.Gross*d.Discount/100
d["Net"] = d.Gross-d.Discount_Amount
d["Age_Group"] = (d.Age>=40).astype(int)
d["Price_Per_Unit"] = d.Gross/d.Quantity
print(d[["Gross","Discount_Amount","Net","Age_Group","Price_Per_Unit"]])
```

## Topic 18 – Perform feature selection using multiple techniques.

### Question / Task
Compare variance threshold and SelectKBest on a small classification example.

### Python Code

```python
X = pd.DataFrame({"A":[0,1,0,1,0,1],"B":[5,5,5,5,5,5],"C":[1,2,1,2,1,2]})
y = [0,1,0,1,0,1]
print("Variance:", X.columns[VarianceThreshold().fit(X).get_support()].tolist())
print("KBest:", X.columns[SelectKBest(f_classif,k=1).fit(X,y).get_support()].tolist())
```
