## Dataset Setup

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
 "Customer_ID":[101,101,102,102,103,103,104,104],
 "Product_ID":["P1","P2","P1","P3","P2","P3","P1","P2"],
 "Age":[22,22,35,35,47,47,29,29],
 "Annual_Income":[300000,300000,650000,650000,900000,900000,450000,450000],
 "Product_Price":[1000,2500,1000,4000,2500,4000,1000,2500],
 "Quantity":[2,1,3,1,2,2,1,3],
 "Product_Rating":[4.2,4.8,4.2,4.5,4.8,4.5,4.2,4.8],
 "Purchase_Date":pd.to_datetime(["2026-01-03","2026-02-10","2026-01-09","2026-03-15","2026-02-04","2026-04-12","2026-01-20","2026-05-10"]),
 "Membership_Type":["Basic","Basic","Silver","Silver","Gold","Gold","Basic","Basic"],
 "Repeat_Purchase":[1,1,1,1,1,1,0,0]
})
df["Order_Value"] = df["Product_Price"] * df["Quantity"]
print(df.head())
```

## 1. What are Interaction Features?

### Definition
Interaction features combine two or more inputs to describe a relationship. The combined value can contain information missing from either input alone. This feature-engineering concept should be evaluated against the prediction goal.

### Business Example
Retail teams combine price and quantity to measure transaction value. The example uses customer shopping transactions to make the idea concrete.

### AI/ML Use
Models may use interactions to capture effects that change with another feature. Always check whether the feature is available at prediction time.

### Python Example

```python
print(df[["Product_Price","Quantity","Order_Value"]].head())
```

### Explanation
Order_Value is an interaction of price and quantity. This is a compact example using the same sample dataset.

### When to Use
Use when a business relationship depends on multiple variables. Choose this technique when it matches the data and business goal.

### Limitations
Avoid interactions without a meaningful hypothesis. Check for leakage and test whether the engineered feature actually improves validation performance.

## 5. Feature / Feature

### Definition
Division expresses one quantity relative to another. Denominators must be nonzero. This feature-engineering concept should be evaluated against the prediction goal.

### Business Example
A retailer compares an order to annual income. The example uses customer shopping transactions to make the idea concrete.

### AI/ML Use
Relative spending can capture affordability better than raw amounts. Always check whether the feature is available at prediction time.

### Python Example

```python
df["Spend_to_Income"] = df["Order_Value"] / df["Annual_Income"].replace(0,np.nan)
print(df[["Spend_to_Income"]].head())
```

### Explanation
Order value is divided by annual income; zero incomes become missing rather than infinite. This is a compact example using the same sample dataset.

### When to Use
Use for scale-independent ratios. Choose this technique when it matches the data and business goal.

### Limitations
Tiny denominators can create extreme ratios. Check for leakage and test whether the engineered feature actually improves validation performance.

## 6. Polynomial Features

### Definition
Polynomial terms include powers and optionally interactions. They help represent nonlinear patterns. This feature-engineering concept should be evaluated against the prediction goal.

### Business Example
A business models spending changes that accelerate with age or income. The example uses customer shopping transactions to make the idea concrete.

### AI/ML Use
Polynomial features help linear models learn curved effects. Always check whether the feature is available at prediction time.

### Python Example

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2, include_bias=False)
print(poly.fit_transform(df[["Age","Quantity"]]).shape)
```

### Explanation
Two original inputs produce five features: two originals, two squares, and one interaction. This is a compact example using the same sample dataset.

### When to Use
Use with simple models when nonlinear effects are plausible. Choose this technique when it matches the data and business goal.

### Limitations
Higher degrees rapidly increase feature count and overfitting risk. Check for leakage and test whether the engineered feature actually improves validation performance.

## 7. Pairwise Interactions

### When to Use

### Limitations

## 8. Domain-Based Interactions

### Definition
Domain-based features use subject-matter knowledge to create meaningful relationships. This feature-engineering concept should be evaluated against the prediction goal.

### Business Example
Order value divided by annual income estimates spending burden. The example uses customer shopping transactions to make the idea concrete.

### AI/ML Use
Business-informed features can be more useful than arbitrary combinations. Always check whether the feature is available at prediction time.

### Python Example

```python
df["Order_Share_of_Income_Pct"] = 100 * df["Order_Value"] / df["Annual_Income"]
print(df[["Order_Share_of_Income_Pct"]].head())
```

### Explanation
The result estimates what percentage of annual income one order represents. This is a compact example using the same sample dataset.

### When to Use
Use when a meaningful real-world relationship is known. Choose this technique when it matches the data and business goal.

### Limitations
