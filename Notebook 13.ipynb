## Complete Practical Example
The following example connects the ideas from this notebook.

```python
from sklearn.preprocessing import FunctionTransformer

def create_features(frame):
    frame = frame.copy()
    frame['Order_Value'] = frame['Quantity'] * frame['Product_Price']
    return frame

numeric = ['Age','Annual_Income','Quantity','Product_Price','Order_Value']
categorical = ['City','Membership_Type']
num_pipeline = Pipeline([('impute',SimpleImputer(strategy='median')),('scale',StandardScaler())])
cat_pipeline = Pipeline([('impute',SimpleImputer(strategy='most_frequent')),('encode',OneHotEncoder(handle_unknown='ignore'))])
processor = ColumnTransformer([('num',num_pipeline,numeric),('cat',cat_pipeline,categorical)],sparse_threshold=0)
full_pipeline = Pipeline([
    ('create',FunctionTransformer(create_features,validate=False)),
    ('preprocess',processor),
    ('select',SelectKBest(score_func=f_classif,k=6)),
    ('model',LogisticRegression(max_iter=1000))
])
full_pipeline.fit(X_train,y_train)
pred = full_pipeline.predict(X_test)
print('Train shape:',X_train.shape,'Test shape:',X_test.shape)
print('Selected features:',full_pipeline.named_steps['select'].get_support().sum())
print('Test accuracy:',round(accuracy_score(y_test,pred),3))
print('New customer prediction:',full_pipeline.predict(X_test.head(1)))
```
